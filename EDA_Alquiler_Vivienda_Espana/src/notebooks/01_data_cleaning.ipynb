# 01 — Data Cleaning
EDA — Precio del Alquiler en España

```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('../data/raw/VDP001_01.csv', sep=';')
df.head()
```

```python
# Información general
df.info()
```

```python
# Normalizar columnas
df.columns = [c.strip().upper().replace(' ', '_') for c in df.columns]
df.columns
```

```python
# Convertir tipos
df['AÑO'] = pd.to_numeric(df['AÑO'], errors='coerce')
df['VALOR'] = df['VALOR'].astype(str).str.replace(',', '.')
df['VALOR'] = pd.to_numeric(df['VALOR'], errors='coerce')
```

```python
# Valores nulos
df.isnull().sum()
```

```python
# Duplicados
df.duplicated().sum()
```

```python
# Eliminar duplicados
df = df.drop_duplicates()
```

```python
# Guardar dataset limpio
df.to_csv('../data/processed/VDP001_01_clean.csv', index=False)
```
